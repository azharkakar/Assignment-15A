```python
#Q1: What is the average fare by service type?
#Answer:
df.groupby("service_type")["fare"].mean()
```

```python
#Q2: Which city has the most trips?
#Answer: Islamabad
df.groupby("city").count()
```

```python
#Q4: Compare trip ratings by gender.
#Answer:
df.groupby("gender")["trip_rating"].mean()
```

```python
#Q5: Make a bar chart of service type vs. average fare.
#Answer:
avg_fare = df.groupby('service_type')['fare'].mean()
import matplotlib.pyplot as plt
plt.bar(avg_fare.index, avg_fare.values, color="green", width=0.5)
plt.xlabel("service_type")
plt.ylabel("avg_fare")
plt.title("bar chart of service type vs average fare")
plt.show()
```
